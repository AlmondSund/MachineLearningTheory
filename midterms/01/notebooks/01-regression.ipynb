# 1. Regression in Feature Spaces and Kernel Formulation

Let an i.i.d. dataset

$$
\{(x_n, t_n)\}_{n=1}^{N}, \qquad x_n \in \mathbb{R}^{P}, \quad t_n \in \mathbb{R},
$$

be given, and let

$$
\phi : \mathbb{R}^{P} \to \mathcal{H}
$$

be a mapping into a feature space $\mathcal{H}$, not necessarily finite-dimensional.

Consider the model

$$
t_n = \langle w, \phi(x_n) \rangle_{\mathcal{H}} + \eta_n, \qquad \eta_n \sim \mathcal{N}(0, \sigma_{\eta}^{2}).
$$

---

We introduce the sampling operator

$$
\Phi : \mathcal{H} \to \mathbb{R}^{N},
\qquad
(\Phi w)_n = \langle w, \phi(x_n) \rangle_{\mathcal{H}},
$$

together with its adjoint

$$
\Phi^{\ast} : \mathbb{R}^{N} \to \mathcal{H},
\qquad
\Phi^{\ast} c = \sum_{n=1}^{N} c_n \, \phi(x_n).
$$

Thus the spaces of the relevant objects are

$$
w \in \mathcal{H},
\qquad
t = (t_1, \dots, t_N)^\top \in \mathbb{R}^{N},
$$

$$
\Phi^{\ast}\Phi : \mathcal{H} \to \mathcal{H},
\qquad
\Phi\Phi^{\ast} : \mathbb{R}^{N} \to \mathbb{R}^{N}.
$$

When one chooses coordinates in a finite-dimensional feature space $\mathcal{H} \cong \mathbb{R}^{M}$, the operator $\Phi$ becomes the usual design matrix

$$
\Phi = \begin{bmatrix}
\phi(x_1)^\top \\
\vdots \\
\phi(x_N)^\top
\end{bmatrix} \in \mathbb{R}^{N \times M}.
$$

Hence the Hilbert-space formulation specializes to the standard matrix formulation when the feature representation is finite-dimensional.

---


**1.1.** Formulate the regularized least-squares problem in the feature space $\mathcal{H}$. In your answer, explicitly write the objective function to be minimized, indicate the unknown variable of the problem, and briefly explain the role of the regularization parameter.


We seek the estimator $\widehat w \in \mathcal{H}$ minimizing the regularized empirical quadratic risk

$$
J(w)
= \frac{1}{2}\sum_{n=1}^{N} \bigl(t_n - \langle w, \phi(x_n) \rangle_{\mathcal{H}}\bigr)^2
+ \frac{\lambda}{2} \lVert w \rVert_{\mathcal{H}}^2,
\qquad \lambda > 0.
$$

Equivalently, using $t \in \mathbb{R}^{N}$ and $\Phi : \mathcal{H} \to \mathbb{R}^{N}$,

$$
J(w) = \frac{1}{2} \lVert t - \Phi w \rVert_2^2 + \frac{\lambda}{2} \lVert w \rVert_{\mathcal{H}}^2.
$$

The unknown variable is therefore $w \in \mathcal{H}$.

The regularization parameter $\lambda$ penalizes large norms of $w$ in feature space. This controls the complexity of the predictor, stabilizes the problem when $\Phi^{\ast}\Phi : \mathcal{H} \to \mathcal{H}$ is singular or ill-conditioned, and induces the usual bias-variance tradeoff: larger values of $\lambda$ produce stronger shrinkage toward the origin in $\mathcal{H}$.

---


**1.2.** Starting from the problem formulated in the previous item, obtain its solution. To do so, write the problem in matrix notation, develop the optimality conditions step by step, and arrive at the final expression for the estimator.


We work in the Hilbert space $\mathcal{H}$ and keep the spaces of all objects explicit:

$$
w, h \in \mathcal{H},
\qquad
t \in \mathbb{R}^{N},
\qquad
\Phi : \mathcal{H} \to \mathbb{R}^{N},
\qquad
\Phi^{\ast} : \mathbb{R}^{N} \to \mathcal{H}.
$$

For any direction $h \in \mathcal{H}$,

$$
\frac{d}{d\varepsilon} J(w + \varepsilon h)\Big|_{\varepsilon = 0}
= -\langle t - \Phi w, \Phi h \rangle_{\mathbb{R}^{N}} + \lambda \langle w, h \rangle_{\mathcal{H}}.
$$

Using the adjoint relation $\langle \Phi u, c \rangle_{\mathbb{R}^{N}} = \langle u, \Phi^{\ast} c \rangle_{\mathcal{H}}$, we obtain

$$
DJ(w)[h]
= \bigl\langle -\Phi^{\ast}(t - \Phi w) + \lambda w,\; h \bigr\rangle_{\mathcal{H}}.
$$

Hence the gradient is the element of $\mathcal{H}$ given by

$$
\nabla J(w) = \Phi^{\ast}\Phi w - \Phi^{\ast} t + \lambda w.
$$

The first-order optimality condition $\nabla J(\widehat w) = 0$ yields the normal equation

$$
(\Phi^{\ast}\Phi + \lambda I_{\mathcal{H}}) \widehat w = \Phi^{\ast} t,
$$

where

$$
\Phi^{\ast}\Phi : \mathcal{H} \to \mathcal{H},
\qquad
I_{\mathcal{H}} : \mathcal{H} \to \mathcal{H}.
$$

We now justify invertibility. The operator $\Phi^{\ast}\Phi$ is self-adjoint because

$$
(\Phi^{\ast}\Phi)^{\ast} = \Phi^{\ast}(\Phi^{\ast})^{\ast} = \Phi^{\ast}\Phi,
$$

and it is positive because, for every $u \in \mathcal{H}$,

$$
\langle u, \Phi^{\ast}\Phi u \rangle_{\mathcal{H}} = \lVert \Phi u \rVert_2^2 \ge 0.
$$

Therefore the operator

$$
A := \Phi^{\ast}\Phi + \lambda I_{\mathcal{H}} : \mathcal{H} \to \mathcal{H}
$$

is self-adjoint and strictly positive. Indeed,

$$
\langle u, Au \rangle_{\mathcal{H}} = \lVert \Phi u \rVert_2^2 + \lambda \lVert u \rVert_{\mathcal{H}}^2 \ge \lambda \lVert u \rVert_{\mathcal{H}}^2.
$$

Hence $A$ is bounded below by $\lambda > 0$, so $A$ is injective and has closed range. Since $A$ is self-adjoint,

$$
\operatorname{Ran}(A)^{\perp} = \ker(A^{\ast}) = \ker(A) = \{0\},
$$

which implies that $\operatorname{Ran}(A)$ is dense. Being both closed and dense, it follows that $\operatorname{Ran}(A) = \mathcal{H}$. Thus $A$ is surjective, hence bijective, and its inverse is bounded with

$$
\lVert A^{-1} \rVert \le \frac{1}{\lambda}.
$$

Therefore,

$$
\boxed{\widehat w = (\Phi^{\ast}\Phi + \lambda I_{\mathcal{H}})^{-1} \Phi^{\ast} t.}
$$

This proves existence and uniqueness of the minimizer.

If now $\mathcal{H} \cong \mathbb{R}^{M}$ is finite-dimensional, then the same theory becomes the standard matrix formulation with

$$
\Phi \in \mathbb{R}^{N \times M},
\qquad
w \in \mathbb{R}^{M},
\qquad
t \in \mathbb{R}^{N},
$$

and therefore

$$
\boxed{\widehat w = (\Phi^\top \Phi + \lambda I_M)^{-1} \Phi^\top t.}
$$

This is the finite-dimensional matrix realization of the Hilbert-space solution above.

**1.3.** Consider the polynomial mapping

$$
\phi(x) = \begin{bmatrix} 1 \\ x \\ x^{2} \end{bmatrix}.
$$

For a dataset $\{(x_n, t_n)\}_{n=1}^{N}$, construct the design matrix associated with this mapping and implement in Python the solution obtained in the previous item. Use a simple numerical example to verify the result and briefly interpret the solution obtained.


This polynomial example is a concrete finite-dimensional realization of the general Hilbert-space theory developed in Items 1.1 and 1.2. Here the input must be scalar, so we are specializing the original setting $x_n \in \mathbb{R}^{P}$ to the case $x_n \in \mathbb{R}$.

Since $\phi(x_n) = [1, x_n, x_n^2]^\top \in \mathbb{R}^{3}$, the associated design matrix is

$$
\Phi =
\begin{bmatrix}
1 & x_1 & x_1^2 \\
1 & x_2 & x_2^2 \\
\vdots & \vdots & \vdots \\
1 & x_N & x_N^2
\end{bmatrix} \in \mathbb{R}^{N \times 3},
\qquad
t \in \mathbb{R}^{N},
\qquad
w \in \mathbb{R}^{3}.
$$

Hence the estimator becomes

$$
\widehat w = (\Phi^\top \Phi + \lambda I_3)^{-1} \Phi^\top t.
$$

To verify the expression, consider the simple dataset

$$
x = (-1, 0, 1, 2)^\top,
\qquad
t = (0, 1, 4, 9)^\top.
$$

These values satisfy exactly the quadratic relation $t_n = 1 + 2x_n + x_n^2$. For $\lambda = 0$, exact recovery of the coefficients requires the design matrix to have full column rank. In this example,

$$
\Phi =
\begin{bmatrix}
1 & -1 & 1 \\
1 & 0 & 0 \\
1 & 1 & 1 \\
1 & 2 & 4
\end{bmatrix}
$$

has rank $3$ because the sample points are distinct, so the polynomial coefficients are uniquely identifiable and the estimator should recover $\widehat w = (1, 2, 1)^\top$.

```python
import numpy as np


def ridge_poly2(x: np.ndarray, t: np.ndarray, lam: float) -> tuple[np.ndarray, np.ndarray]:
    """Return ridge coefficients and the design matrix for phi(x) = [1, x, x^2]."""
    x = np.asarray(x, dtype=float).reshape(-1)
    t = np.asarray(t, dtype=float).reshape(-1)

    if x.shape != t.shape:
        raise ValueError("x and t must have the same shape")
    if lam < 0:
        raise ValueError("lam must be nonnegative")

    Phi = np.column_stack([np.ones_like(x), x, x**2])
    w = np.linalg.solve(Phi.T @ Phi + lam * np.eye(3), Phi.T @ t)
    return w, Phi


def predict_poly2(x: np.ndarray, w: np.ndarray) -> np.ndarray:
    x = np.asarray(x, dtype=float).reshape(-1)
    Phi = np.column_stack([np.ones_like(x), x, x**2])
    return Phi @ w
```

Interpretation. For $\lambda = 0$, the model recovers exactly the coefficients of the underlying quadratic polynomial and interpolates the sample. For $\lambda > 0$, the coefficients are shrunk toward the origin, so the fit is no longer exact; this is precisely the regularizing effect of ridge regression. To make this effect visually explicit, it is useful to sweep $\lambda$ across several orders of magnitude and observe how the fitted curve moves from an almost interpolating regime to a much more strongly constrained regime.

The following animation shows the fitted curve as $\lambda$ varies from $0$ to large positive values. The left panel tracks the regression function, while the right panel keeps the design matrix $\Phi$ explicit and updates the ridge system $(\Phi^\top \Phi + \lambda I) \widehat w = \Phi^\top t$ together with the resulting vectors $\widehat w(\lambda)$ and $\Phi \widehat w(\lambda)$.


```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML


def format_matrix_block(name: str, values: np.ndarray) -> str:
    values = np.asarray(values, dtype=float)
    body = np.array2string(
        values,
        precision=2,
        suppress_small=True,
        floatmode="fixed",
        max_line_width=72,
    )
    return f"{name} =\n{body}"


def animate_lambda_sweep(x: np.ndarray, t: np.ndarray, num_frames: int = 80) -> HTML:
    x = np.asarray(x, dtype=float).reshape(-1)
    t = np.asarray(t, dtype=float).reshape(-1)

    if x.shape != t.shape:
        raise ValueError("x and t must have the same shape")

    _, Phi = ridge_poly2(x, t, lam=0.0)
    gram = Phi.T @ Phi
    rhs = Phi.T @ t
    x_grid = np.linspace(x.min() - 0.5, x.max() + 0.5, 400)
    lam_values = np.concatenate(([0.0], np.logspace(-4, 2, num_frames - 1)))

    fig = plt.figure(figsize=(11, 4.8))
    grid = fig.add_gridspec(1, 2, width_ratios=[1.9, 1.1], wspace=0.18)
    ax_curve = fig.add_subplot(grid[0, 0])
    ax_info = fig.add_subplot(grid[0, 1])

    ax_curve.scatter(x, t, color="black", zorder=3, label="training data")
    fit_line, = ax_curve.plot([], [], color="tab:blue", lw=2.5, label="ridge fit")
    reference, = ax_curve.plot(
        x_grid,
        1 + 2 * x_grid + x_grid**2,
        "--",
        color="tab:orange",
        lw=1.5,
        label="reference",
    )
    label = ax_curve.text(0.03, 0.95, "", transform=ax_curve.transAxes, va="top")

    ax_curve.set_xlim(x_grid.min(), x_grid.max())
    y_pad = 0.1 * (t.max() - t.min() + 1.0)
    ax_curve.set_ylim(t.min() - y_pad, t.max() + y_pad)
    ax_curve.set_xlabel("x")
    ax_curve.set_ylabel("prediction")
    ax_curve.set_title("Effect of the Regularization Parameter")
    ax_curve.legend(loc="lower right")
    ax_curve.grid(alpha=0.2)

    ax_info.set_title(r"Design-matrix view of $(\Phi^\top \Phi + \lambda I) \widehat w = \Phi^\top t$", fontsize=11)
    ax_info.set_facecolor("#f7f7f7")
    ax_info.set_xticks([])
    ax_info.set_yticks([])
    for spine in ax_info.spines.values():
        spine.set_visible(False)
    info_text = ax_info.text(
        0.0,
        0.98,
        "",
        va="top",
        ha="left",
        family="monospace",
        fontsize=8.5,
        transform=ax_info.transAxes,
    )

    def update(frame: int):
        lam = float(lam_values[frame])
        w, _ = ridge_poly2(x, t, lam=lam)
        y_grid = predict_poly2(x_grid, w)
        y_train = Phi @ w
        system = gram + lam * np.eye(Phi.shape[1])

        fit_line.set_data(x_grid, y_grid)
        if lam == 0.0:
            label.set_text(r"$\lambda = 0$")
        else:
            label.set_text(rf"$\lambda = {lam:.2e}$")

        info_text.set_text(
            "\n\n".join(
                [
                    format_matrix_block(r"$\Phi$", Phi),
                    format_matrix_block(r"$\Phi^\top t$", rhs),
                    format_matrix_block(r"$\Phi^\top \Phi + \lambda I$", system),
                    format_matrix_block(r"$\widehat w(\lambda)$", w),
                    format_matrix_block(r"$\Phi \widehat w(\lambda)$", y_train),
                ]
            )
        )
        return fit_line, reference, label, info_text

    update(0)
    animation = FuncAnimation(fig, update, frames=len(lam_values), interval=200, blit=False)
    plt.close(fig)
    return HTML(animation.to_jshtml(default_mode="loop"))


x = np.array([-1.0, 0.0, 1.0, 2.0])
t = np.array([0.0, 1.0, 4.0, 9.0])


animate_lambda_sweep(x, t)
```

**1.3.** Kernel formulation from the SVD of the design operator. The printed statement repeats the label 1.3; we follow that numbering here.


Although $\mathcal{H}$ may be infinite-dimensional, the subspace

$$
\mathcal{S} := \operatorname{span}\{\phi(x_1), \dots, \phi(x_N)\} \subseteq \mathcal{H}
$$

has dimension $m \le N$. Choosing an orthonormal basis of $\mathcal{S}$, the restriction of the sampling operator to this finite-dimensional subspace admits a matrix representation

$$
\Phi \in \mathbb{R}^{N \times m},
$$

so the following SVD argument is performed on the finite-rank coordinate representation of the transformed sample span, not on the whole of $\mathcal{H}$. In particular, every matrix identity written below is an identity on the coordinate spaces $\mathbb{R}^{m}$ and $\mathbb{R}^{N}$ attached to this active sample-generated subspace.

Let the compact singular value decomposition be

$$
\Phi = U \Sigma V^\top,
$$

where

$$
U \in \mathbb{R}^{N \times r},
\qquad
\Sigma = \operatorname{diag}(\sigma_1, \dots, \sigma_r) \in \mathbb{R}^{r \times r},
\qquad
V \in \mathbb{R}^{m \times r},
$$

and $r = \operatorname{rank}(\Phi)$. Then

$$
\Phi^\top \Phi = V \Sigma^2 V^\top : \mathbb{R}^{m} \to \mathbb{R}^{m},
$$

$$
\Phi \Phi^\top = U \Sigma^2 U^\top : \mathbb{R}^{N} \to \mathbb{R}^{N}.
$$

Thus $\Phi^\top \Phi$ acts in feature coordinates, whereas $\Phi \Phi^\top$ acts in sample coordinates. Both operators have the same nonzero eigenvalues $\sigma_1^2, \dots, \sigma_r^2$.

We now derive the key algebraic identity explicitly. Since

$$
\Phi^\top \Phi = V \Sigma^2 V^\top,
\qquad
\Phi^\top = V \Sigma U^\top,
$$

we have

$$
(\Phi^\top \Phi + \lambda I_m)^{-1} \Phi^\top
= V (\Sigma^2 + \lambda I_r)^{-1} V^\top V \Sigma U^\top
= V (\Sigma^2 + \lambda I_r)^{-1} \Sigma U^\top.
$$

Because $\Sigma$ is diagonal, it commutes with $(\Sigma^2 + \lambda I_r)^{-1}$, so

$$
V (\Sigma^2 + \lambda I_r)^{-1} \Sigma U^\top
= V \Sigma (\Sigma^2 + \lambda I_r)^{-1} U^\top.
$$

On the other hand,

$$
\Phi \Phi^\top = U \Sigma^2 U^\top,
$$

hence

$$
\Phi^\top (\Phi \Phi^\top + \lambda I_N)^{-1}
= V \Sigma U^\top U (\Sigma^2 + \lambda I_r)^{-1} U^\top
= V \Sigma (\Sigma^2 + \lambda I_r)^{-1} U^\top.
$$

Therefore,

$$
\boxed{(\Phi^\top \Phi + \lambda I_m)^{-1} \Phi^\top = \Phi^\top (\Phi \Phi^\top + \lambda I_N)^{-1}.}
$$

Now define the Gram matrix in sample space:

$$
K := \Phi \Phi^\top \in \mathbb{R}^{N \times N}.
$$

Its entries are

$$
K_{ij} = (\Phi \Phi^\top)_{ij} = \langle \phi(x_i), \phi(x_j) \rangle_{\mathcal{H}} =: k(x_i, x_j).
$$

Thus the kernel matrix $K$ records all pairwise inner products between transformed samples without requiring explicit access to the coordinates of $\phi(x_i)$.

Starting from the primal estimator in feature coordinates,

$$
\widehat w = (\Phi^\top \Phi + \lambda I_m)^{-1} \Phi^\top t,
$$

the identity proved above yields

$$
\widehat w = \Phi^\top (K + \lambda I_N)^{-1} t.
$$

If we define the sample-space coefficient vector

$$
\alpha := (K + \lambda I_N)^{-1} t \in \mathbb{R}^{N},
$$

then $\widehat w = \Phi^\top \alpha$ and, for any new point $x$,

$$
\widehat f(x)
= \langle \widehat w, \phi(x) \rangle_{\mathcal{H}}
= \sum_{n=1}^{N} \alpha_n \, \langle \phi(x_n), \phi(x) \rangle_{\mathcal{H}}
= \sum_{n=1}^{N} \alpha_n \, k(x_n, x).
$$

Consequently, the primal formulation manipulates $\Phi^\top \Phi$ and the feature-space parameter $w$, whereas the kernelized formulation manipulates $K = \Phi \Phi^\top$ and the sample-space parameter $\alpha$. Passing from $w$ to $\alpha$ is the dual reformulation that makes kernelization possible; the kernel trick proper is the subsequent replacement of explicit feature-space inner products $\langle \phi(x_i), \phi(x_j) \rangle_{\mathcal{H}}$ by kernel evaluations $k(x_i, x_j)$. In this way one computes entirely through the Gram matrix entries $K_{ij} = k(x_i, x_j)$, without constructing the feature vectors themselves.

```python
_, Phi = ridge_poly2(x, t, lam=0.0)

sv_feature = np.linalg.svd(Phi.T @ Phi, compute_uv=False)
sv_sample = np.linalg.svd(Phi @ Phi.T, compute_uv=False)
sv_phi_squared = np.linalg.svd(Phi, compute_uv=False) ** 2
nonzero_sv_sample = sv_sample[sv_sample > 1e-12]

print("Singular values of Phi.T @ Phi:")
print(np.array2string(sv_feature, precision=6, suppress_small=True))
print()
print("Singular values of Phi @ Phi.T:")
print(np.array2string(sv_sample, precision=6, suppress_small=True))
print()
print("Squared singular values of Phi:")
print(np.array2string(sv_phi_squared, precision=6, suppress_small=True))
print()
print("Nonzero singular values match:", np.allclose(sv_feature, nonzero_sv_sample))
print("Extra singular values in Phi @ Phi.T:", np.array2string(sv_sample[len(sv_feature):], precision=6, suppress_small=True))
```
