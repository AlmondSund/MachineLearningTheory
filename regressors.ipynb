```python
import warnings
from typing import Optional

import numpy as np


def _validate_snr_db(
    snr_db: float,  # Requested signal-to-noise ratio [dB]
) -> float:        # Validated SNR in decibels
    '''Validates and normalizes the requested SNR in decibels.'''
    # Convert to a scalar to reject non-numeric or array-like inputs early.
    try:
        snr_db_value = float(snr_db)
    except (TypeError, ValueError) as exc:
        raise ValueError("snr_db must be a real scalar in decibels [dB].") from exc

    if np.isnan(snr_db_value):
        raise ValueError("snr_db must not be NaN.")

    return snr_db_value


def _is_quantizing_dtype(
    dtype: np.dtype,  # Candidate dtype for the output signal
) -> bool:           # True when casting back would quantize AWGN heavily
    '''Returns whether preserving this dtype would quantize or erase AWGN.'''
    return np.issubdtype(dtype, np.integer) or np.issubdtype(dtype, np.bool_)


def _resolve_output_dtype(
    x_dtype: np.dtype,      # Input dtype
    is_complex: bool,       # Whether the signal is complex-valued
    preserve_dtype: bool,   # Whether output must keep the input dtype
) -> np.dtype:              # Chosen output dtype for y = x + n
    '''Resolves the output dtype according to preserve_dtype semantics.'''
    if preserve_dtype:
        return np.dtype(x_dtype)
    return np.dtype(np.complex128 if is_complex else np.float64)


def _compute_signal_power(
    x_t: np.ndarray,  # Input signal samples (real or complex)
) -> float:          # Average discrete-time signal power E[|x|^2]
    '''Computes signal power robustly using promoted precision.

    Parameters
    ----------
    x_t : np.ndarray
        Input signal samples.

    Returns
    -------
    float
        Average signal power E[|x|^2] as a finite non-negative scalar.

    Side Effects
    ------------
    None.
    '''
    # Promote before multiplication to prevent overflow for integer inputs.
    if np.iscomplexobj(x_t):
        x_for_power = x_t.astype(np.complex128, copy=False)
        # Use x*conj(x) to avoid abs/exponent temporaries for complex inputs.
        power_samples = (x_for_power * np.conjugate(x_for_power)).real
    else:
        x_for_power = x_t.astype(np.float64, copy=False)
        # Use x*x to avoid abs/exponent temporaries for real inputs.
        power_samples = x_for_power * x_for_power

    signal_power = float(np.mean(power_samples, dtype=np.float64))
    if not np.isfinite(signal_power):
        raise ValueError("x_t power is non-finite after dtype promotion; check signal magnitude.")
    return signal_power


def _compute_noise_std(
    signal_power: float,  # Signal power E[|x|^2]
    snr_db: float,        # Requested signal-to-noise ratio [dB]
) -> float:               # AWGN standard deviation for real-valued components
    '''Converts SNR [dB] to a validated AWGN standard deviation.

    Parameters
    ----------
    signal_power : float
        Average signal power E[|x|^2].
    snr_db : float
        Requested SNR in decibels [dB].

    Returns
    -------
    float
        Standard deviation used for real AWGN components.

    Side Effects
    ------------
    None.
    '''
    # Derive linear SNR and noise power while keeping overflow/underflow silent.
    with np.errstate(over="ignore", under="ignore", divide="ignore", invalid="ignore"):
        snr_linear = np.power(10.0, snr_db / 10.0)
        noise_power = np.divide(signal_power, snr_linear)
        noise_std = np.sqrt(noise_power)

    # Validate derived parameters while allowing legitimate zero-noise limits.
    if not np.isfinite(noise_power) or noise_power < 0.0:
        raise ValueError("Derived noise_power is invalid; snr_db is too extreme for stable generation.")
    if not np.isfinite(noise_std) or noise_std < 0.0:
        raise ValueError("Derived noise_std is invalid; snr_db is too extreme for stable generation.")

    return float(noise_std)


def _sample_standard_normal(
    shape: tuple[int, ...],   # Desired shape of random samples
    dtype: np.dtype,          # Floating dtype for the samples
    rng: np.random.Generator, # Random generator used for sampling
) -> np.ndarray:              # Standard normal samples N(0, 1)
    '''Samples N(0, 1) with a preferred dtype while keeping compatibility.

    Parameters
    ----------
    shape : tuple[int, ...]
        Target shape.
    dtype : np.dtype
        Preferred floating dtype for generated samples.
    rng : np.random.Generator
        Generator from which random numbers are consumed.

    Returns
    -------
    np.ndarray
        Standard normal samples with the requested shape and practical dtype.

    Side Effects
    ------------
    Consumes random numbers from ``rng``.
    '''
    preferred_dtype = np.dtype(dtype)
    if preferred_dtype == np.float64:
        return rng.standard_normal(size=shape)

    # Use dtype-aware generation when available, then fall back to casting.
    try:
        return rng.standard_normal(size=shape, dtype=preferred_dtype)
    except TypeError:
        return rng.standard_normal(size=shape).astype(preferred_dtype, copy=False)


def _draw_white_noise(
    shape: tuple[int, ...],   # Desired shape for the sampled noise
    noise_std: float,         # Standard deviation for real noise components
    is_complex: bool,         # Whether to generate complex-valued noise
    rng: np.random.Generator, # Random generator used to sample noise
    output_dtype: np.dtype,   # Final dtype for returned noise samples
) -> np.ndarray:              # White Gaussian noise samples
    '''Draws white Gaussian noise in the requested output dtype.

    Parameters
    ----------
    shape : tuple[int, ...]
        Target shape.
    noise_std : float
        Standard deviation for real-valued components.
    is_complex : bool
        True for complex AWGN generation.
    rng : np.random.Generator
        Generator from which random numbers are consumed.
    output_dtype : np.dtype
        Final dtype for the returned noise array.

    Returns
    -------
    np.ndarray
        White Gaussian noise with the requested shape.

    Side Effects
    ------------
    Consumes random numbers from ``rng``.
    '''
    output_dtype_obj = np.dtype(output_dtype)

    # For complex AWGN each component gets half the total variance.
    if is_complex:
        component_dtype = np.float32 if output_dtype_obj == np.complex64 else np.float64
        component_std = np.asarray(noise_std / np.sqrt(2.0), dtype=component_dtype)
        noise_real = _sample_standard_normal(shape=shape, dtype=component_dtype, rng=rng) * component_std
        noise_imag = _sample_standard_normal(shape=shape, dtype=component_dtype, rng=rng) * component_std
        noise = noise_real + 1j * noise_imag
        return noise.astype(output_dtype_obj, copy=False)

    component_dtype = np.float32 if output_dtype_obj == np.float16 else output_dtype_obj
    scale = np.asarray(noise_std, dtype=component_dtype)
    noise_real = _sample_standard_normal(shape=shape, dtype=component_dtype, rng=rng) * scale
    return noise_real.astype(output_dtype_obj, copy=False)


def awgn(
    x_t: np.ndarray,                            # Input signal samples (real or complex)
    snr_db: float,                              # Target SNR with respect to mean signal power [dB]
    rng: Optional[np.random.Generator] = None,  # RNG for reproducible noise generation
    preserve_dtype: bool = True,                # If True, cast output back to x_t.dtype
) -> np.ndarray:                                # Noisy signal with the same shape as x_t
    '''Adds additive white Gaussian noise (AWGN) to a discrete-time signal.

    Purpose
    -------
    Perturbs a real or complex signal with zero-mean white Gaussian noise so
    that the target signal-to-noise ratio matches ``snr_db``.

    Parameters
    ----------
    x_t : np.ndarray
        Input signal samples. Power is estimated as E[|x|^2] on the array.
    snr_db : float
        Desired signal-to-noise ratio in decibels [dB].
        ``np.inf`` means no noise is added.
    rng : Optional[np.random.Generator], default=None
        Random number generator used for noise sampling.
    preserve_dtype : bool, default=True
        If True, cast output back to ``x_t.dtype``.
        If False, return ``float64`` for real inputs and ``complex128`` for
        complex inputs. For finite ``snr_db``, integer/bool inputs with
        ``preserve_dtype=True`` are rejected to prevent quantized AWGN.
        ``float16`` is allowed but may heavily quantize low-amplitude noise,
        so a runtime warning is emitted in that configuration.
        ``complex64`` is allowed; at extreme finite SNR values, limited
        dynamic range can increase overflow/saturation risk.

    Returns
    -------
    np.ndarray
        Noisy signal with the same shape as ``x_t``.

    Side Effects
    ------------
    Consumes random numbers from ``rng`` or from a newly created generator.

    Assumptions
    -----------
    - ``x_t`` contains at least one sample.
    - ``x_t`` has non-zero average power unless ``snr_db == np.inf``.
    - ``snr_db`` is real and not NaN.
    - ``snr_db = -np.inf`` is invalid because it implies infinite noise power.
    '''
    x = np.asarray(x_t)
    snr_db_value = _validate_snr_db(snr_db)
    is_complex = np.iscomplexobj(x)
    output_dtype = _resolve_output_dtype(x.dtype, is_complex, preserve_dtype)

    # Guard against degenerate inputs before power and SNR calculations.
    if x.size == 0:
        raise ValueError("x_t must contain at least one sample.")
    if np.isneginf(snr_db_value):
        raise ValueError("snr_db = -inf is not supported because it implies infinite noise power.")
    if np.isposinf(snr_db_value):
        if preserve_dtype:
            return x.copy()
        return x.astype(output_dtype, copy=True)
    if preserve_dtype and _is_quantizing_dtype(x.dtype):
        raise ValueError(
            "preserve_dtype=True with integer/bool x_t would quantize AWGN. "
            "Use preserve_dtype=False or cast x_t to float/complex first."
        )
    if preserve_dtype and np.dtype(x.dtype) == np.float16:
        warnings.warn(
            "preserve_dtype=True with float16 can heavily quantize AWGN; "
            "consider preserve_dtype=False for better noise fidelity.",
            category=RuntimeWarning,
            stacklevel=2,
        )

    # Estimate average signal power with the standard discrete-time definition.
    signal_power = _compute_signal_power(x)
    if signal_power == 0.0:
        raise ValueError("x_t has zero power; SNR is undefined for a null signal.")

    # Convert SNR from decibels and validate derived AWGN parameters.
    noise_std = _compute_noise_std(signal_power=signal_power, snr_db=snr_db_value)

    # If derived noise is exactly zero, return without consuming RNG state.
    if noise_std == 0.0:
        if np.dtype(x.dtype) == output_dtype:
            return x.copy()
        return x.astype(output_dtype, copy=True)
    rng_obj = rng if rng is not None else np.random.default_rng()

    # Draw AWGN directly in the final output dtype.
    noise = _draw_white_noise(
        shape=x.shape,
        noise_std=noise_std,
        is_complex=is_complex,
        rng=rng_obj,
        output_dtype=output_dtype,
    )

    # Avoid a redundant cast when input and output dtypes already match.
    if np.dtype(x.dtype) == output_dtype:
        return x + noise
    return x.astype(output_dtype, copy=False) + noise
```

```python
# Create a pure sinusoid x(t).
sample_rate_hz = 1000  # Sampling frequency [Hz]
duration_s = 1.0       # Signal duration [s]
tone_freq_hz = 5.0     # Sinusoid frequency [Hz]
amplitude = 1.0        # Sinusoid amplitude [-]

t_s = np.arange(0.0, duration_s, 1.0 / sample_rate_hz, dtype=np.float64)  # Time samples [s]
x_t = amplitude * np.sin(2.0 * np.pi * tone_freq_hz * t_s)                                                           # Signal sample
```

```python
# Build a synthetic AWGN dataset from the same clean sinusoid x_t.
snr_values_db = [1.0, 2.0, 5.0, 10.0]       # Target SNR levels [dB]
rng_dataset = np.random.default_rng(seed=2026)

dataset_awgn = []
for snr_db in snr_values_db:
    # Generate one noisy realization y_t for each SNR level.
    y_t = awgn(x_t=x_t, snr_db=snr_db, rng=rng_dataset, preserve_dtype=True)

    # Store clean, noisy, and residual noise signals together.
    dataset_awgn.append(
        {
            'snr_db': snr_db,
            'x_t': x_t.copy(),
            'y_t': y_t,
            'n_t': y_t - x_t,
        }
    )

# Optional array view for ML workflows: shape = (num_snr, num_samples).
X_clean = np.stack([entry['x_t'] for entry in dataset_awgn], axis=0)
Y_noisy = np.stack([entry['y_t'] for entry in dataset_awgn], axis=0)

print(f'Dataset entries: {len(dataset_awgn)}')
print(f'X_clean shape: {X_clean.shape}, Y_noisy shape: {Y_noisy.shape}')
```

```python
from dataclasses import dataclass, field
from typing import Any, Dict, List, Optional, Tuple

import matplotlib.pyplot as plt
import numpy as np
from IPython.display import HTML
from matplotlib.animation import FuncAnimation
from numpy.polynomial.legendre import legvander


@dataclass
class FitResult:
    '''Stores one regression fit and its uncertainty model.'''
    method: str
    beta_hat: np.ndarray
    sigma2_hat: float
    cov_beta: np.ndarray
    extras: Dict[str, Any] = field(default_factory=dict)


def build_polynomial_design_matrix(
    t_s: np.ndarray,                     # Time samples [s]
    q_dim: int,                          # Number of polynomial basis functions Q
    normalize_time: bool = True,         # If True, map time to [-1, 1] before polynomial expansion
    basis_family: str = 'legendre',      # Polynomial basis family: 'legendre' or 'monomial'
) -> np.ndarray:                         # Polynomial design matrix Phi with shape (N, Q)
    '''Builds a polynomial design matrix with selectable polynomial basis family.

    Purpose:
        Construct a polynomial basis that is linear in parameters and reusable by
        OLS, Ridge, ML, and MAP estimators.

    Parameters:
        t_s: Time-domain samples [s].
        q_dim: Number of polynomial columns (must be >= 1).
        normalize_time: Whether to scale the input time vector to [-1, 1].
        basis_family: Polynomial family for feature construction:
            - 'monomial': [1, t, t^2, ..., t^(Q-1)]
            - 'legendre': [P0(t), P1(t), ..., P(Q-1)(t)] on [-1, 1]

    Returns:
        A matrix Phi of shape (N, Q), where N is the number of samples.

    Side effects:
        None.
    '''
    t_vector = np.asarray(t_s, dtype=np.float64).reshape(-1)

    if t_vector.size == 0:
        raise ValueError('t_s must contain at least one sample.')
    if q_dim < 1:
        raise ValueError('q_dim must be at least 1.')
    if basis_family not in {'monomial', 'legendre'}:
        raise ValueError("basis_family must be either 'monomial' or 'legendre'.")

    # Keep polynomial powers numerically stable by operating on a bounded domain.
    if normalize_time:
        t_min = float(np.min(t_vector))
        t_max = float(np.max(t_vector))
        if t_max <= t_min:
            raise ValueError('t_s must contain at least two distinct values when normalize_time=True.')
        t_for_basis = 2.0 * (t_vector - t_min) / (t_max - t_min) - 1.0
    else:
        t_for_basis = t_vector

    # Use orthogonal Legendre polynomials by default for better conditioning.
    if basis_family == 'legendre':
        # legvander(x, deg) returns columns [P0(x), ..., Pdeg(x)].
        return legvander(t_for_basis, int(q_dim) - 1)

    # increasing=True generates monomial columns [t^0, t^1, ..., t^(Q-1)].
    return np.vander(t_for_basis, N=int(q_dim), increasing=True)


def split_train_test_indices(
    num_samples: int,                           # Number of total samples N
    train_fraction: float = 0.8,                # Fraction assigned to train set [-]
    rng: Optional[np.random.Generator] = None,  # RNG for reproducible shuffling
) -> Tuple[np.ndarray, np.ndarray]:             # (idx_train, idx_test)
    '''Creates a reproducible random split of sample indices into train/test sets.

    Purpose:
        Build a stable evaluation protocol where all methods use the same split.

    Parameters:
        num_samples: Number of available samples.
        train_fraction: Proportion assigned to training.
        rng: Optional generator to control randomness.

    Returns:
        Two sorted integer arrays with train and test indices.

    Side effects:
        Consumes random numbers from ``rng`` when provided.
    '''
    if num_samples < 2:
        raise ValueError('num_samples must be at least 2.')
    if not (0.0 < train_fraction < 1.0):
        raise ValueError('train_fraction must be strictly between 0 and 1.')

    rng_obj = rng if rng is not None else np.random.default_rng(seed=0)
    perm = rng_obj.permutation(num_samples)

    n_train = int(np.floor(train_fraction * num_samples))
    n_train = int(np.clip(n_train, 1, num_samples - 1))

    # Sort to keep plotting consistent with increasing time order.
    idx_train = np.sort(perm[:n_train])
    idx_test = np.sort(perm[n_train:])
    return idx_train, idx_test


def solve_linear_system(
    lhs_matrix: np.ndarray,  # Left-hand matrix A in A x = b
    rhs_array: np.ndarray,   # Right-hand side b
) -> np.ndarray:             # Solution x
    '''Solves A x = b with a least-squares fallback when A is singular.

    Purpose:
        Keep estimators robust in ill-conditioned regimes without crashing.

    Parameters:
        lhs_matrix: Coefficient matrix A.
        rhs_array: Right-hand side vector or matrix b.

    Returns:
        Exact solution when possible; least-squares solution otherwise.

    Side effects:
        None.
    '''
    try:
        return np.linalg.solve(lhs_matrix, rhs_array)
    except np.linalg.LinAlgError:
        solution, *_ = np.linalg.lstsq(lhs_matrix, rhs_array, rcond=None)
        return solution


def gaussian_nll(
    beta: np.ndarray,   # Regression parameters beta
    sigma2: float,      # Noise variance sigma^2 (> 0)
    phi: np.ndarray,    # Design matrix Phi
    y: np.ndarray,      # Observed targets y
) -> float:             # Negative log-likelihood value
    '''Computes Gaussian negative log-likelihood for y ~ N(Phi beta, sigma^2 I).

    Purpose:
        Provide a likelihood metric for ML fitting and diagnostics.

    Parameters:
        beta: Parameter vector.
        sigma2: Observation noise variance.
        phi: Design matrix.
        y: Observed target vector.

    Returns:
        Scalar negative log-likelihood value.

    Side effects:
        None.
    '''
    if sigma2 <= 0.0:
        raise ValueError('sigma2 must be strictly positive for Gaussian NLL.')

    residual = y - phi @ beta
    n_samples = y.size
    squared_error = float(residual @ residual)

    return float(
        0.5 * n_samples * np.log(2.0 * np.pi * sigma2)
        + 0.5 * squared_error / sigma2
    )


def fit_ols(
    phi_train: np.ndarray,  # Training design matrix Phi_train
    y_train: np.ndarray,    # Training targets y_train
) -> FitResult:             # OLS fit result
    '''Fits ordinary least squares and returns uncertainty diagnostics.

    Purpose:
        Estimate polynomial coefficients that minimize ||y - Phi beta||^2.

    Parameters:
        phi_train: Training design matrix.
        y_train: Training target samples.

    Returns:
        ``FitResult`` containing beta_hat, ML-style residual variance, and
        covariance proxy Cov(beta_hat) = sigma2_hat * (Phi^T Phi)^(-1).

    Side effects:
        None.
    '''
    gram = phi_train.T @ phi_train
    rhs = phi_train.T @ y_train

    # Solve normal equations while tolerating singular Gram matrices.
    beta_hat = solve_linear_system(lhs_matrix=gram, rhs_array=rhs)

    # Use ML normalization (1/N) for residual variance estimation.
    residual = y_train - phi_train @ beta_hat
    sigma2_hat = float((residual @ residual) / y_train.size)

    # Propagate residual variance to parameter-space uncertainty.
    gram_inv = solve_linear_system(lhs_matrix=gram, rhs_array=np.eye(gram.shape[0], dtype=np.float64))
    cov_beta = sigma2_hat * gram_inv

    return FitResult(
        method='OLS',
        beta_hat=beta_hat,
        sigma2_hat=sigma2_hat,
        cov_beta=cov_beta,
    )


def fit_ridge(
    phi_train: np.ndarray,   # Training design matrix Phi_train
    y_train: np.ndarray,     # Training targets y_train
    lambda_l2: float = 1e-3, # L2 regularization weight lambda >= 0
) -> FitResult:              # Ridge fit result
    '''Fits Ridge regression (Tikhonov regularization).

    Purpose:
        Stabilize polynomial coefficient estimates under collinearity by
        penalizing large parameter magnitudes.

    Parameters:
        phi_train: Training design matrix.
        y_train: Training target samples.
        lambda_l2: Non-negative L2 penalty.

    Returns:
        ``FitResult`` with ridge coefficients, residual variance estimate,
        and covariance approximation.

    Side effects:
        None.
    '''
    if lambda_l2 < 0.0:
        raise ValueError('lambda_l2 must be non-negative.')

    gram = phi_train.T @ phi_train
    rhs = phi_train.T @ y_train
    system = gram + float(lambda_l2) * np.eye(gram.shape[0], dtype=np.float64)

    # Solve the regularized normal equations directly.
    beta_hat = solve_linear_system(lhs_matrix=system, rhs_array=rhs)

    residual = y_train - phi_train @ beta_hat
    sigma2_hat = float((residual @ residual) / y_train.size)

    # Use the ridge covariance approximation based on A^{-1} G A^{-1}.
    system_inv = solve_linear_system(lhs_matrix=system, rhs_array=np.eye(system.shape[0], dtype=np.float64))
    cov_beta = sigma2_hat * (system_inv @ gram @ system_inv)

    return FitResult(
        method='Ridge',
        beta_hat=beta_hat,
        sigma2_hat=sigma2_hat,
        cov_beta=cov_beta,
        extras={'lambda_l2': float(lambda_l2)},
    )


def fit_ml(
    phi_train: np.ndarray,                            # Training design matrix Phi_train
    y_train: np.ndarray,                              # Training targets y_train
    beta_ols_reference: Optional[np.ndarray] = None,  # OLS beta for equivalence check
    sigma2_fixed_for_beta: float = 1.0,              # Fixed variance for argmin_beta(NLL)
) -> FitResult:                                       # ML fit result
    '''Fits Gaussian maximum likelihood parameters.

    Purpose:
        Show that minimizing Gaussian NLL over beta with fixed sigma^2 yields
        the same beta as OLS, and estimate sigma^2 from residuals.

    Parameters:
        phi_train: Training design matrix.
        y_train: Training target samples.
        beta_ols_reference: Optional OLS coefficients used for equivalence check.
        sigma2_fixed_for_beta: Positive value used in beta-only NLL minimization.

    Returns:
        ``FitResult`` with ML beta, sigma^2, covariance estimate, and metadata
        containing NLL and whether beta matches OLS.

    Side effects:
        None.
    '''
    if sigma2_fixed_for_beta <= 0.0:
        raise ValueError('sigma2_fixed_for_beta must be strictly positive.')

    gram = phi_train.T @ phi_train
    rhs = phi_train.T @ y_train

    # The scalar 1/(2*sigma^2) does not change the minimizer in beta.
    beta_hat = solve_linear_system(lhs_matrix=gram, rhs_array=rhs)

    if beta_ols_reference is None:
        beta_ols_reference = fit_ols(phi_train=phi_train, y_train=y_train).beta_hat
    beta_matches_ols = bool(np.allclose(beta_hat, beta_ols_reference, rtol=1e-10, atol=1e-10))

    residual = y_train - phi_train @ beta_hat
    sigma2_hat = float((residual @ residual) / y_train.size)

    gram_inv = solve_linear_system(lhs_matrix=gram, rhs_array=np.eye(gram.shape[0], dtype=np.float64))
    cov_beta = sigma2_hat * gram_inv

    nll_value = gaussian_nll(beta=beta_hat, sigma2=sigma2_hat, phi=phi_train, y=y_train)

    return FitResult(
        method='ML',
        beta_hat=beta_hat,
        sigma2_hat=sigma2_hat,
        cov_beta=cov_beta,
        extras={
            'nll': float(nll_value),
            'beta_matches_ols': beta_matches_ols,
            'sigma2_fixed_for_beta': float(sigma2_fixed_for_beta),
        },
    )


def fit_map(
    phi_train: np.ndarray,   # Training design matrix Phi_train
    y_train: np.ndarray,     # Training targets y_train
    sigma2: float,           # Likelihood noise variance sigma^2 > 0
    tau2: float = 1.0,       # Prior variance beta ~ N(0, tau^2 I), tau^2 > 0
) -> FitResult:              # MAP fit result
    '''Fits MAP for linear-Gaussian regression with isotropic Gaussian prior.

    Purpose:
        Compute closed-form MAP estimate and posterior covariance to capture
        coefficient shrinkage induced by the prior.

    Parameters:
        phi_train: Training design matrix.
        y_train: Training target samples.
        sigma2: Observation noise variance.
        tau2: Prior variance of beta.

    Returns:
        ``FitResult`` with MAP coefficients and posterior covariance.

    Side effects:
        None.
    '''
    if sigma2 <= 0.0:
        raise ValueError('sigma2 must be strictly positive.')
    if tau2 <= 0.0:
        raise ValueError('tau2 must be strictly positive.')

    gram = phi_train.T @ phi_train
    rhs = phi_train.T @ y_train

    # MAP mean is equivalent to ridge with lambda = sigma2 / tau2.
    regularization = float(sigma2 / tau2)
    system = gram + regularization * np.eye(gram.shape[0], dtype=np.float64)
    beta_hat = solve_linear_system(lhs_matrix=system, rhs_array=rhs)

    # Posterior covariance is the inverse posterior precision matrix.
    posterior_precision = gram / sigma2 + np.eye(gram.shape[0], dtype=np.float64) / tau2
    cov_beta = solve_linear_system(
        lhs_matrix=posterior_precision,
        rhs_array=np.eye(posterior_precision.shape[0], dtype=np.float64),
    )

    return FitResult(
        method='MAP',
        beta_hat=beta_hat,
        sigma2_hat=float(sigma2),
        cov_beta=cov_beta,
        extras={'tau2': float(tau2)},
    )


def predict_mean_and_std(
    phi: np.ndarray,                       # Design matrix for prediction points
    beta_hat: np.ndarray,                  # Estimated coefficients beta_hat
    cov_beta: Optional[np.ndarray] = None, # Optional covariance of beta_hat
) -> Tuple[np.ndarray, Optional[np.ndarray]]: # (predictive_mean, predictive_std)
    '''Predicts mean output and optional parameter-induced predictive std.

    Purpose:
        Produce estimated clean-signal trajectories (x-hat) and uncertainty
        bands from fitted regression parameters.

    Parameters:
        phi: Design matrix at prediction coordinates.
        beta_hat: Estimated coefficient vector.
        cov_beta: Optional covariance matrix of coefficients.

    Returns:
        Tuple with mean prediction and standard deviation. Std is ``None`` when
        covariance is not provided.

    Side effects:
        None.
    '''
    x_hat = phi @ beta_hat
    if cov_beta is None:
        return x_hat, None

    # Propagate parameter covariance into prediction space: phi_i^T Cov(beta) phi_i.
    predictive_var = np.einsum('ij,jk,ik->i', phi, cov_beta, phi)
    predictive_var = np.maximum(predictive_var, 0.0)
    predictive_std = np.sqrt(predictive_var)
    return x_hat, predictive_std


def evaluate_methods_for_split(
    phi_all: np.ndarray,        # Design matrix for all samples (N, Q)
    y_all: np.ndarray,          # Noisy observations (N,)
    x_all: np.ndarray,          # Clean reference used for test MSE (N,)
    idx_train: np.ndarray,      # Training indices
    idx_test: np.ndarray,       # Test indices
    lambda_l2: float,           # Ridge regularization weight
    tau2_prior: float,          # MAP prior variance tau^2
) -> Dict[str, Dict[str, Any]]: # Per-method fit data, predictions, and metrics
    '''Fits all estimators on one split and computes prediction metrics.

    Purpose:
        Centralize one consistent evaluation pass for OLS, Ridge, ML, and MAP
        so static plotting, sweep plotting, and animation reuse the same logic.

    Parameters:
        phi_all: Full design matrix.
        y_all: Full noisy signal vector.
        x_all: Full clean signal vector.
        idx_train: Training sample indices.
        idx_test: Test sample indices.
        lambda_l2: Ridge penalty weight.
        tau2_prior: MAP prior variance.

    Returns:
        Dictionary keyed by method name with fit object, full predictions,
        predictive std, and test MSE against the clean signal.

    Side effects:
        None.
    '''
    phi_train = phi_all[idx_train]
    phi_test = phi_all[idx_test]
    y_train = y_all[idx_train]
    x_test = x_all[idx_test]

    # Fit all methods on the same training set to ensure fair comparisons.
    fit_ols_result = fit_ols(phi_train=phi_train, y_train=y_train)
    fit_ridge_result = fit_ridge(phi_train=phi_train, y_train=y_train, lambda_l2=lambda_l2)
    fit_ml_result = fit_ml(
        phi_train=phi_train,
        y_train=y_train,
        beta_ols_reference=fit_ols_result.beta_hat,
        sigma2_fixed_for_beta=1.0,
    )
    fit_map_result = fit_map(
        phi_train=phi_train,
        y_train=y_train,
        sigma2=fit_ols_result.sigma2_hat,
        tau2=tau2_prior,
    )

    all_fits = [fit_ols_result, fit_ridge_result, fit_ml_result, fit_map_result]
    results: Dict[str, Dict[str, Any]] = {}

    # Compute clean-signal estimates on test/full grid and test MSE vs x_t.
    for fit_result in all_fits:
        xhat_test, _ = predict_mean_and_std(
            phi=phi_test,
            beta_hat=fit_result.beta_hat,
            cov_beta=fit_result.cov_beta,
        )
        xhat_all, xhat_std_all = predict_mean_and_std(
            phi=phi_all,
            beta_hat=fit_result.beta_hat,
            cov_beta=fit_result.cov_beta,
        )

        mse_test = float(np.mean((xhat_test - x_test) ** 2))

        results[fit_result.method] = {
            'fit': fit_result,
            'xhat_all': xhat_all,
            'xhat_std_all': xhat_std_all,
            'mse_test': mse_test,
        }

    return results


# Experiment configuration.
q_sweep_values = np.arange(1, 31)   # Candidate polynomial basis sizes Q
ridge_lambda = 1e-3                 # Ridge regularization weight
map_tau2 = 1.0                      # MAP prior variance tau^2
animation_interval_ms = 180         # Animation frame interval [ms]
polynomial_basis_family = 'legendre'  # Polynomial basis used for design matrix construction

method_names = ['OLS', 'Ridge', 'ML', 'MAP']
method_colors = {
    'OLS': 'tab:orange',
    'Ridge': 'tab:green',
    'ML': 'tab:red',
    'MAP': 'tab:purple',
}
method_line_styles = {
    'OLS': '-',
    'Ridge': '-',
    'ML': '--',
    'MAP': '-.',
}
method_line_widths = {
    'OLS': 1.5,
    'Ridge': 1.4,
    'ML': 1.2,
    'MAP': 1.2,
}


# Convert dataset entries to a typed, float64-centric structure.
entries_by_snr: Dict[float, Dict[str, np.ndarray]] = {}
for entry in dataset_awgn:
    snr_db = float(entry['snr_db'])
    entries_by_snr[snr_db] = {
        'x_t': np.asarray(entry['x_t'], dtype=np.float64),
        'y_t': np.asarray(entry['y_t'], dtype=np.float64),
    }


# Reuse one deterministic train/test split per SNR for all Q values.
split_rng = np.random.default_rng(seed=12345)
split_indices_by_snr: Dict[float, Tuple[np.ndarray, np.ndarray]] = {}
for snr_db in sorted(entries_by_snr):
    num_samples = entries_by_snr[snr_db]['y_t'].size
    idx_train, idx_test = split_train_test_indices(
        num_samples=num_samples,
        train_fraction=0.8,
        rng=split_rng,
    )
    split_indices_by_snr[snr_db] = (idx_train, idx_test)


# Sweep polynomial dimension Q and cache predictions for animation.
sweep_rows: List[Dict[str, Any]] = []
animation_cache: Dict[int, Dict[float, Dict[str, np.ndarray]]] = {}

for q_dim in q_sweep_values:
    q_value = int(q_dim)
    phi_q = build_polynomial_design_matrix(
        t_s=t_s,
        q_dim=q_value,
        normalize_time=True,
        basis_family=polynomial_basis_family,
    )
    animation_cache[q_value] = {}

    for snr_db in sorted(entries_by_snr):
        entry = entries_by_snr[snr_db]
        idx_train, idx_test = split_indices_by_snr[snr_db]

        method_results = evaluate_methods_for_split(
            phi_all=phi_q,
            y_all=entry['y_t'],
            x_all=entry['x_t'],
            idx_train=idx_train,
            idx_test=idx_test,
            lambda_l2=ridge_lambda,
            tau2_prior=map_tau2,
        )

        # Keep only the x-hat trajectories needed by the animation.
        animation_cache[q_value][snr_db] = {
            method_name: method_results[method_name]['xhat_all']
            for method_name in method_names
        }

        for method_name in method_names:
            sweep_rows.append(
                {
                    'snr_db': snr_db,
                    'method': method_name,
                    'q_dim': q_value,
                    'mse_test': float(method_results[method_name]['mse_test']),
                }
            )


# Build MSE lookup tables and best-Q summaries.
sweep_mse_lookup: Dict[Tuple[float, str, int], float] = {
    (row['snr_db'], row['method'], row['q_dim']): row['mse_test']
    for row in sweep_rows
}

best_q_by_snr_method: Dict[Tuple[float, str], Dict[str, Any]] = {}
best_q_overall_by_snr: Dict[float, Dict[str, Any]] = {}

for snr_db in sorted(entries_by_snr):
    rows_for_snr = [row for row in sweep_rows if row['snr_db'] == snr_db]
    best_q_overall_by_snr[snr_db] = min(rows_for_snr, key=lambda row: row['mse_test'])

    for method_name in method_names:
        rows_for_method = [
            row for row in rows_for_snr
            if row['method'] == method_name
        ]
        best_q_by_snr_method[(snr_db, method_name)] = min(rows_for_method, key=lambda row: row['mse_test'])


# Build one combined figure:
# - Top row: animated signal reconstructions versus Q.
# - Bottom row: MSE-versus-Q curves with a moving current-Q marker.
snr_values_sorted = sorted(entries_by_snr)
fig_combo, axes_combo = plt.subplots(2, len(snr_values_sorted), figsize=(22, 9), dpi=100, sharex='row')
axes_signal = axes_combo[0]
axes_mse = axes_combo[1]

anim_signal_handles: Dict[Tuple[float, str], Any] = {}
anim_mse_marker_handles: Dict[Tuple[float, str], Any] = {}
anim_mse_qline_handles: Dict[float, Any] = {}

first_q = int(q_sweep_values[0])
for col_idx, snr_db in enumerate(snr_values_sorted):
    ax_signal = axes_signal[col_idx]
    ax_mse = axes_mse[col_idx]
    entry = entries_by_snr[snr_db]

    # Label only on the first column to avoid repeated legend entries.
    observed_label = r'Observed $y_t$' if col_idx == 0 else '_nolegend_'
    clean_label = r'Clean $x_t$' if col_idx == 0 else '_nolegend_'

    ax_signal.plot(
        t_s,
        entry['y_t'],
        color='tab:blue',
        linewidth=1.0,
        alpha=0.65,
        label=observed_label,
    )
    ax_signal.plot(
        t_s,
        entry['x_t'],
        color='black',
        linestyle='--',
        linewidth=1.2,
        label=clean_label,
    )

    for method_name in method_names:
        xhat_initial = animation_cache[first_q][snr_db][method_name]
        method_label = fr'{method_name} $\hat{{x}}_t$' if col_idx == 0 else '_nolegend_'
        line_handle, = ax_signal.plot(
            t_s,
            xhat_initial,
            color=method_colors[method_name],
            linestyle=method_line_styles[method_name],
            linewidth=method_line_widths[method_name],
            label=method_label,
        )
        anim_signal_handles[(snr_db, method_name)] = line_handle

    ax_signal.set_title(f'SNR = {snr_db:.0f} dB | Q = {first_q}')
    ax_signal.set_xlabel('Time [s]')
    ax_signal.set_ylabel('Amplitude [-]')
    ax_signal.set_xlim(float(np.min(t_s)), float(np.max(t_s)))
    ax_signal.grid(True, alpha=0.3)

    # Plot full MSE curves and keep a moving marker to indicate current Q.
    for method_name in method_names:
        mse_curve = [
            sweep_mse_lookup[(snr_db, method_name, int(q_dim))]
            for q_dim in q_sweep_values
        ]

        mse_label = fr'{method_name} $\mathrm{{MSE}}(\hat{{x}}_t, x_t)$' if col_idx == 0 else '_nolegend_'
        ax_mse.plot(
            q_sweep_values,
            mse_curve,
            color=method_colors[method_name],
            linestyle=method_line_styles[method_name],
            linewidth=1.3,
            marker='o',
            markersize=3,
            label=mse_label,
        )

        best_method_row = best_q_by_snr_method[(snr_db, method_name)]
        ax_mse.scatter(
            best_method_row['q_dim'],
            best_method_row['mse_test'],
            color=method_colors[method_name],
            s=30,
        )

        mse_current = sweep_mse_lookup[(snr_db, method_name, first_q)]
        current_marker_handle, = ax_mse.plot(
            [first_q],
            [mse_current],
            marker='o',
            linestyle='None',
            markersize=7,
            color=method_colors[method_name],
            label='_nolegend_',
        )
        anim_mse_marker_handles[(snr_db, method_name)] = current_marker_handle

    best_overall = best_q_overall_by_snr[snr_db]
    best_label = r'Overall best $\mathrm{MSE}(\hat{x}_t, x_t)$' if col_idx == 0 else '_nolegend_'
    ax_mse.scatter(
        best_overall['q_dim'],
        best_overall['mse_test'],
        color='black',
        marker='*',
        s=140,
        label=best_label,
    )

    q_line_handle = ax_mse.axvline(
        x=first_q,
        color='gray',
        linestyle='--',
        linewidth=1.0,
        alpha=0.8,
        label='_nolegend_',
    )
    anim_mse_qline_handles[snr_db] = q_line_handle

    ax_mse.set_title(f'SNR = {snr_db:.0f} dB | MSE vs Q')
    ax_mse.set_xlabel('Polynomial dimension Q [-]')
    ax_mse.set_ylabel(r'Test $\mathrm{MSE}(\hat{x}_t, x_t)$ [-]')
    ax_mse.set_xlim(float(np.min(q_sweep_values)), float(np.max(q_sweep_values)))
    ax_mse.grid(True, alpha=0.3)

# Place row legends on the rightmost subplot of each row.
signal_handles, signal_labels = axes_signal[0].get_legend_handles_labels()
mse_handles, mse_labels = axes_mse[0].get_legend_handles_labels()

axes_signal[-1].legend(
    signal_handles,
    signal_labels,
    loc='upper right',
    fontsize=8,
    frameon=True,
)

axes_mse[-1].legend(
    mse_handles,
    mse_labels,
    loc='upper right',
    fontsize=7,
    frameon=True,
)

fig_combo.tight_layout()


def _update_q_animation(
    frame_index: int,  # Frame position in q_sweep_values
) -> List[Any]:        # Updated artists list
    '''Updates estimator curves for one animation frame.

    Purpose:
        Rebind each method trajectory to the current polynomial dimension Q.

    Parameters:
        frame_index: Integer frame index inside ``q_sweep_values``.

    Returns:
        A list of modified matplotlib artists for the frame.

    Side effects:
        Mutates line artists and subplot titles in ``fig_combo``.
    '''
    q_current = int(q_sweep_values[frame_index])
    updated_artists: List[Any] = []

    for col_idx, snr_db in enumerate(snr_values_sorted):
        ax_signal = axes_signal[col_idx]
        ax_mse = axes_mse[col_idx]

        # Update reconstructed clean-signal traces for the current Q.
        for method_name in method_names:
            line_handle = anim_signal_handles[(snr_db, method_name)]
            line_handle.set_ydata(animation_cache[q_current][snr_db][method_name])
            updated_artists.append(line_handle)

            # Update the current-Q marker on each MSE curve.
            mse_current = sweep_mse_lookup[(snr_db, method_name, q_current)]
            marker_handle = anim_mse_marker_handles[(snr_db, method_name)]
            marker_handle.set_data([q_current], [mse_current])
            updated_artists.append(marker_handle)

        # Update the vertical guide line that indicates current Q.
        q_line_handle = anim_mse_qline_handles[snr_db]
        q_line_handle.set_xdata([q_current, q_current])
        updated_artists.append(q_line_handle)

        ax_signal.set_title(f'SNR = {snr_db:.0f} dB | Q = {q_current}')
        ax_mse.set_title(f'SNR = {snr_db:.0f} dB | MSE vs Q (Q = {q_current})')

    return updated_artists


q_sweep_animation = FuncAnimation(
    fig=fig_combo,
    func=_update_q_animation,
    frames=len(q_sweep_values),
    interval=animation_interval_ms,
    blit=False,
    repeat=True,
)

# Render the animation inline in notebook output.
# Increase the inline embed limit so large animations are not truncated.
plt.rcParams['animation.embed_limit'] = max(float(plt.rcParams['animation.embed_limit']), 80.0)
q_sweep_animation_html = HTML(q_sweep_animation.to_jshtml())
# Avoid rendering the static first animation frame as an extra plot.
plt.close(fig_combo)
q_sweep_animation_html
```
